## 3.3. Dos fidelidades, y por qué no es Successive Halving

En Random Forest y en la EBM la búsqueda evalúa cada configuración con un presupuesto bajo (100 árboles; 2 bolsas
externas) y solo la ganadora se reajusta con el presupuesto alto (300 árboles; 8 bolsas): un factor de 3 y de 4.

Eso **no** es Successive Halving. Successive Halving parte de $n$ configuraciones con un presupuesto pequeño, conserva
la fracción $1/r$ mejor, multiplica el presupuesto por $r$ y repite hasta que queda una; $r$ es el **factor de
reducción**, habitualmente 3, y Hyperband repite el procedimiento con varias parejas de $n$ y presupuesto inicial. En el
esquema del proyecto no hay rondas intermedias ni configuraciones que compitan con el presupuesto alto: es una búsqueda
con dos fidelidades. La comparación lado a lado, con las mismas configuraciones, está en el
[dashboard](https://fuga-de-clientes-dash.onrender.com) (Contexto › «Dos fidelidades, no Successive Halving»).
