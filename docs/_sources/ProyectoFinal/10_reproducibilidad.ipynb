## 10.3. Cómo reproducir cada resultado

Desde `ModeloOriginal/`, con los CSV de COFINFAD en `datos/` (el repositorio los trae comprimidos y se extraen solos),
la {ref}`Tabla 10.3 <tab-reproducir>` dice qué guion produce cada resultado.

(tab-reproducir)=
**Tabla 10.3.** Guiones del proyecto final, en el orden en que se ejecutan, y lo que produce cada uno.

| Paso | Orden | Produce |
|---|---|---|
| Pruebas unitarias | `python -m pytest -q` | Verifica el modelo, el marco de experimentos, las pruebas de hipótesis y el informe |
| Búsqueda de literatura | `python experiments/00_busqueda_literatura.py` | `resultados/literatura/` (capítulo 4) |
| Auditoría de fuga | `python experiments/01_auditoria_fuga.py` | `resultados/auditoria_fuga.json` (sección 2.5) |
| Validación anidada | `python experiments/02_validacion_anidada.py --semillas 42 43 44` | `resultados/experimentos.parquet`, la tabla maestra (capítulos 3 y 6) |
| Evaluación en prueba | `python experiments/03_prueba.py` | `predicciones_prueba.parquet`, `metricas_prueba.csv`, `incertidumbre_prueba.csv` |
| Estabilidad por semilla y cómputo | `python experiments/03b_semillas.py` | `semillas.csv` |
| Memoria e inferencia según las filas | `python experiments/03c_memoria.py` | `memoria.csv`, `inferencia_filas.csv` (secciones 7.3 y 7.5) |
| Métodos recientes con hiperparámetros recomendados | `python experiments/03d_recomendados.py` | `recomendados.csv` (sección 3.6) |
| Techo teórico | `python experiments/04b_techo.py` | `techo.json` (sección 6.4) |
| Pruebas de hipótesis | `python experiments/04_pruebas_hipotesis.py` | `pruebas_hipotesis.csv`, `veredicto.json`, `pliegues_<tarea>.csv` |
| Predictor óptimo | `python experiments/04c_optimo.py` | `optimo.json` (sección 6.9) |
| Empates de la campaña | `python experiments/04d_empates_campana.py` | `empates_campana.json` (sección 6.9) |
| Referencias homogéneas | `python experiments/04e_referencias_homogeneas.py` | `referencias_homogeneas.json` (secciones 6.7 y 7.8) |
| Conjuntos adicionales | `python experiments/05_conjuntos_adicionales.py` | `resultados/adicionales/<conjunto>/` (sección 7.1) |
| Robustez | `python experiments/06_robustez.py` | `resultados/robustez/` (secciones 7.2 a 7.6), incluida la curva de aprendizaje sin la mezcla |
| Ablación | `python experiments/07_ablacion.py` | `ablacion.csv` (sección 5.9) |
| Rutas relativas | `python experiments/09_rutas_relativas.py` | Reescribe las rutas de las tablas maestras relativas al repositorio, para publicarlas |
| Tablas y figuras del libro | `python experiments/08_tablas_libro.py` | `ProyectoFinal/_generado/` y `ProyectoFinal/figuras/` |
| Todo lo anterior, en dos colas | `python experiments/correr_todo.py` y `python experiments/correr_todo.py --cola b` | Ejecuta los pasos en orden y salta lo ya hecho |
| Capturas del dashboard | `python ProyectoFinal/capturar_dash.py --url <dirección del dashboard>` (en la raíz del repositorio) | `ProyectoFinal/figuras/dash_*.png` (capítulo 8) |
| Cuadernos del libro | `python ProyectoFinal/construir.py` (en la raíz del repositorio) | Los capítulos ejecutados, `ProyectoFinal/*.ipynb` |
| Libro | `jupyter-book build .` (en la raíz del repositorio) | `_build/html`, que se publica en `docs/` |
| Informe en PDF | `python ProyectoFinal/exportar_pdf.py` | `ProyectoFinal/ProyectoFinal_informe.pdf` |
