## 4.5. Declaración de novedad

Al mejor de nuestro conocimiento, y según la búsqueda descrita (tres bases de datos, trece cadenas, búsqueda por
citación en los dos sentidos y revisión de repositorios, en las fechas que registra la matriz), **no se encontraron
estudios que apliquen a la predicción de fuga de clientes un modelo de boosting con una mezcla latente de
desplazamientos ajustado por EM generalizado**, ni otro algoritmo de la familia de EM con paso de boosting, ni que
deriven de un modelo así la probabilidad de superar un umbral de riesgo.

La novedad que se afirma es precisa y acotada:

- **Qué no es nuevo.** Sustituir el paso M del EM por un paso de boosting ya se había hecho: en mezclas con exceso de
  ceros {cite}`zhou2022`, en modelos con estados latentes {cite}`adam2022` y en mezclas para pérdidas de seguros, con
  el nombre de *Expectation-Boosting* {cite}`hou2025`. BDL es un miembro de esa familia, no su origen.
- **Qué es propio de BDL.** (i) El modelo: un índice único aprendido por boosting de árboles, con desplazamientos
  latentes constantes y pesos que no dependen de los predictores. Es la contraparte con clase latente de los árboles
  con efectos aleatorios, y un caso mucho más restringido que una mezcla de regresiones. (ii) La forma que toma el
  paso M con ese modelo: un solo paso de boosting contra el objetivo al que se le resta el desplazamiento esperado,
  en lugar de un boosting por componente. (iii) La regla de salida: la probabilidad de excedencia de la mezcla como
  clasificador de una respuesta continua dicotomizada. (iv) La densidad de la mezcla como medida de conformidad de los
  conjuntos de predicción. (v) La implementación, programada desde cero, y su aplicación a la fuga de clientes.
- **Qué se reutiliza.** El algoritmo EM {cite}`dempster1977`, el boosting de gradiente {cite}`friedman2001`, las
  mezclas finitas {cite}`mclachlan2000`, la predicción conforme por partición {cite}`vovk2005,lei2018` y el criterio
  BIC {cite}`schwarz1978`. Ninguna de esas piezas es nuestra.
- **Qué no se afirma.** No se afirma que el modelo estadístico del error sea nuevo (es el de los interceptos aleatorios
  con puntos de masa), ni que la alternancia entre EM y boosting lo sea, ni que BDL sea mejor fuera de los problemas
  con un desplazamiento latente: sin él se reduce a un boosting de mínimos cuadrados (capítulo 7).

Los antecedentes algorítmicos más cercanos son el algoritmo *Expectation-Boosting* de {cite:t}`hou2025`, la regresión de
mezcla con boosting de {cite:t}`jobst2024` y el EM con aprendizaje automático de {cite:t}`xue2024`; la
{ref}`Tabla 4.3 <tab-cercanos>` precisa las diferencias. El mecanismo está formulado de manera general (capítulo 5):
sirve para una regresión en la que una variable discreta que no está entre los predictores desplaza la respuesta, y
para clasificar una respuesta continua dicotomizada. Tiene condiciones: el desplazamiento debe ser aditivo, no depender
de los predictores y ser grande frente al error del índice. En los dos conjuntos adicionales construidos con una
variable omitida esas condiciones no se dieron y el modelo no encontró clases latentes (sección 7.1).
